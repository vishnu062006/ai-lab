```python

# First Order Logic - Unification Algorithm

def is_variable(x):
    return isinstance(x, str) and x.startswith("?")

def substitute(term, subs):
    if isinstance(term, str):
        while term in subs:
            term = subs[term]
        return term

    return (term[0], tuple(substitute(a, subs) for a in term[1]))

def occurs_check(var, term, subs):
    term = substitute(term, subs)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs_check(var, a, subs) for a in term[1])

    return False

def unify(x, y, subs=None):
    if subs is None:
        subs = {}

    x = substitute(x, subs)
    y = substitute(y, subs)

    if x == y:
        return subs

    if is_variable(x):
        if occurs_check(x, y, subs):
            return None
        subs[x] = y
        return subs

    if is_variable(y):
        if occurs_check(y, x, subs):
            return None
        subs[y] = x
        return subs

    if isinstance(x, tuple) and isinstance(y, tuple):
        if x[0] != y[0] or len(x[1]) != len(y[1]):
            return None

        for a, b in zip(x[1], y[1]):
            subs = unify(a, b, subs)

            if subs is None:
                return None

        return subs

    return None


def test_unification(expr1, expr2):
    print("\nExpression 1:", expr1)
    print("Expression 2:", expr2)

    result = unify(expr1, expr2)

    if result is None:
        print("Unification Failed!")
    else:
        print("Unification Successful!")
        print("Substitutions:", result)


# Example 1: Two Variables
print("\n--- Example 1: Two Variables ---")
test_unification(
    ("Eats", ("?x", "?y")),
    ("Eats", ("Riya", "Apple"))
)

# Example 2: Three Variables
print("\n--- Example 2: Three Variables ---")
test_unification(
    ("Knows", ("?x", "?y", "?z")),
    ("Knows", ("Ram", "Sita", "John"))
)

# Example 3: Unification Failure
print("\n--- Example 3: Unification Failure ---")
test_unification(
    ("Knows", ("John", "?x")),
    ("Knows", ("Mary", "Apple"))
)
```
